Example 1: K-Means Clustering


```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
data = pd.read_excel("/content/Mall Customers.xlsx")
X = data[['Annual Income (k$)', 'Spending Score (1-100)']].iloc[:100]
kmeans = KMeans(n_clusters=5, random_state=0)
y_kmeans = kmeans.fit_predict(X)
plt.scatter(X.iloc[:,0], X.iloc[:,1], c=y_kmeans)
plt.scatter(kmeans.cluster_centers_[:,0],
            kmeans.cluster_centers_[:,1],
            s=200,
            marker='X')
plt.xlabel("Annual Income")
plt.ylabel("Spending Score")
plt.title("Customer Segmentation using K-Means")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from kmodes.kmodes import KModes
data = pd.read_excel("/content/Mall Customers.xlsx")
data['AgeGroup'] = pd.cut(data['Age'],
                          bins=[0,25,40,60,100],
                          labels=['Young','Adult','MiddleAge','Senior'])
X = data[['Gender','AgeGroup']]
X = X.astype(str)
km = KModes(n_clusters=3, init='Huang', n_init=5, verbose=1)
clusters = km.fit_predict(X)
data['Cluster'] = clusters
print(data[['Gender','Age','AgeGroup','Cluster']].head())
```

Example 3: Hierarchical Clustering using Single Linkage

```python
# Using Single linkage
import pandas as pd
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import dendrogram, linkage
data = pd.read_excel("/content/Mall Customers.xlsx")
X = data[['Annual Income (k$)', 'Spending Score (1-100)']]
X = X.iloc[:100]
Z = linkage(X, method='single')
plt.figure(figsize=(8,5))
dendrogram(Z)
plt.title("Dendrogram")
plt.xlabel("Customers")
plt.ylabel("Distance")
plt.show()
```

Exercise 1: Apply clustering algorithms to segment customers based on
purchase patterns.

• K-Means Clustering

• Hierarchical Clustering

Dataset: https://www.kaggle.com/code/saurabhbadole/wholesalecustomer-purchasing-behavior

```python
# K-Means Clustering
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
data = pd.read_csv("/content/Wholesale customers data.csv")
X = data[['Fresh','Milk','Grocery','Frozen','Detergents_Paper','Delicassen']]
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
kmeans = KMeans(n_clusters=3, random_state=0)
clusters = kmeans.fit_predict(X_scaled)
data['Cluster'] = clusters
plt.scatter(data['Milk'], data['Grocery'], c=clusters)
plt.xlabel("Milk Spending")
plt.ylabel("Grocery Spending")
plt.title("Customer Segmentation using K-Means")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
data = pd.read_csv("Wholesale customers data.csv")
X = data[['Milk','Grocery']]
kmeans = KMeans(n_clusters=3, random_state=0)
y_kmeans = kmeans.fit_predict(X)
plt.scatter(X.iloc[:,0], X.iloc[:,1], c=y_kmeans)
plt.scatter(kmeans.cluster_centers_[:,0],
            kmeans.cluster_centers_[:,1],
            s=200,
            marker='X')
plt.xlabel("Milk")
plt.ylabel("Grocery")
plt.title("Customer Segmentation using K-Means")
plt.show()
```

```python
# Hierarchical Clustering
import pandas as pd
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import dendrogram, linkage
from sklearn.preprocessing import StandardScaler
data = pd.read_csv("/content/Wholesale customers data.csv")
X = data[['Fresh','Milk','Grocery','Frozen','Detergents_Paper','Delicassen']]
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
Z = linkage(X_scaled, method='ward')
plt.figure(figsize=(10,5))
dendrogram(Z)
plt.title("Dendrogram for Customer Segmentation")
plt.xlabel("Customers")
plt.ylabel("Distance")
plt.show()
```

• Use climate data and group regions based on similar climate

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
data = pd.read_csv("/content/rainfall in india 1901-2015.csv")
yearly_rainfall = data.groupby("YEAR")["ANNUAL"].mean().reset_index()
X = yearly_rainfall[['ANNUAL']]
kmeans = KMeans(n_clusters=3, random_state=0)
yearly_rainfall['Cluster'] = kmeans.fit_predict(X)
plt.scatter(yearly_rainfall['YEAR'],
            yearly_rainfall['ANNUAL'],
            c=yearly_rainfall['Cluster'])
plt.xlabel("Year")
plt.ylabel("Annual Rainfall")
plt.title("Clustering Years Based on Rainfall Similarity")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
data = pd.read_csv("/content/RF_AI_1901-2021.csv")
print(data.head())
X = data[['JUN','JUL','AUG','SEP']]
kmeans = KMeans(n_clusters=3, random_state=0)
data['Cluster'] = kmeans.fit_predict(X)
print(data[['YEAR','JUN','JUL','AUG','SEP','JUN-SEP','Cluster']])
plt.figure(figsize=(10,5))
plt.scatter(data['YEAR'], data['JUN-SEP'], c=data['Cluster'])
centroids = kmeans.cluster_centers_
plt.scatter(data['YEAR'].mean(), centroids[:,0].mean(), s=200, marker='X')
plt.xlabel("Year")
plt.ylabel("Total Monsoon Rainfall (JUN-SEP)")
plt.title("Clustering Years Based on Rainfall Similarity")
plt.show()
```

```python
import requests
import pandas as pd
url = "https://cckpapi.worldbank.org/api/v1/cmip6-x0.25_timeseries_tas,hurs,pr_timeseries_monthly_1950-2014,2015-2100_mean_historical_access-cm2_r1i1p1f1_mean/IND?_format=json"
data = requests.get(url).json()
records = []
for date, value in data['data']['pr']['1950-2014']['IND'].items():
    year, month = date.split("-")
    records.append({
        "Year": int(year),
        "Month": int(month),
        "Rainfall": value
    })
df = pd.DataFrame(records)
df.to_csv("india_rainfall_clean.csv", index=False)
print(df.head())
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
data = pd.read_csv("/content/india_rainfall_clean.csv")
X = data[['Rainfall']]
kmeans = KMeans(n_clusters=3, random_state=0, n_init=10)
data['Cluster'] = kmeans.fit_predict(X)
print(data[['Year','Month','Rainfall','Cluster']].head())
plt.scatter(data['Year'], data['Rainfall'], c=data['Cluster'])
plt.xlabel("Year")
plt.ylabel("Rainfall")
plt.title("Climate Region Clustering (based on Rainfall)")
plt.show()
```
