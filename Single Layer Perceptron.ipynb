Example - 1: SLP for AND gate with learning curve and decision boundary

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import Perceptron 
X = np.array([[0,0],[0,1],[1,0],[1,1]])
y = np.array([0,0,0,1])
w = np.zeros(2)
b = 0
eta = 0.1
errors_per_epoch=[]
epochs=10
for epoch in range(10):
    total_error = 0
    for i in range(len(X)):
        z = np.dot(X[i], w) + b
        y_pred = 1 if z >= 0 else 0
        error = y[i] - y_pred
        w = w + eta * error * X[i]
        b = b + eta * error
        total_error += abs(error)
        print("Weights:", w)
        print("Bias:", b)
    errors_per_epoch.append(total_error)
plt.plot(range(1, epochs+1),errors_per_epoch)
plt.xlabel("Epochs")
plt.ylabel("Total Error")
plt.title("Learning Curve of Single Layer Perceptron")
plt.show()
slp = Perceptron(max_iter=1000, eta0=0.1, 
random_state=1)
slp.fit(X, y)
x_min, x_max = -0.5, 1.5
y_min, y_max = -0.5, 1.5
xx, yy = np.meshgrid(np.linspace(x_min, x_max, 300),
np.linspace(y_min, y_max, 300))
Z = slp.predict(np.c_[xx.ravel(), yy.ravel()])
Z = Z.reshape(xx.shape)
plt.contourf(xx, yy, Z, alpha=0.3)
plt.scatter(X[:, 0], X[:, 1])
plt.xlabel("x1")
plt.ylabel("x2")
plt.title("Decision Boundary of SLP for AND Problem")
plt.show()
```

Exercise - 1: SLP for OR gate with learning curve and decision boundary

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import Perceptron 
X = np.array([[0,0], [0,1], [1,0], [1,1]])
y = np.array([0,1,1,1])
np.random.seed(0)
w = np.random.randn(2)
b = np.random.randn()
learning_rate = 0.1
epochs = 20
errors = []
def step(x):
    return 1 if x >= 0 else 0
for epoch in range(epochs):
    total_error = 0
    for i in range(len(X)):
        linear_output = np.dot(X[i], w) + b
        y_pred = step(linear_output)
        error = y[i] - y_pred
        total_error += abs(error)
        w += learning_rate * error * X[i]
        b += learning_rate * error
    errors.append(total_error)
print("Final Weights:", w)
print("Final Bias:", b)
plt.figure()
plt.plot(range(epochs), errors)
plt.title("Learning Curve (OR Gate)")
plt.xlabel("Epochs")
plt.ylabel("Total Error")
plt.show()
plt.figure()
slp = Perceptron(max_iter=1000, eta0=0.1, random_state=1)
slp.fit(X, y)
x_min, x_max = -0.5, 1.5
y_min, y_max = -0.5, 1.5
xx, yy = np.meshgrid(np.linspace(x_min, x_max, 300),
np.linspace(y_min, y_max, 300))
Z = slp.predict(np.c_[xx.ravel(), yy.ravel()])
Z = Z.reshape(xx.shape)
plt.contourf(xx, yy, Z, alpha=0.3)
plt.scatter(X[:, 0], X[:, 1])
plt.xlabel("x1")
plt.ylabel("x2")
plt.title("Decision Boundary of SLP for OR Problem")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import accuracy_score
data = load_iris()
X = data.data[:, :2]  
y = data.target
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=1
)
def plot_decision_boundary(model, X, y, title):
    x_min, x_max = X[:, 0].min() - 1, X[:, 0].max() + 1
    y_min, y_max = X[:, 1].min() - 1, X[:, 1].max() + 1
    xx, yy = np.meshgrid(
        np.linspace(x_min, x_max, 300),
        np.linspace(y_min, y_max, 300)
    )
    Z = model.predict(np.c_[xx.ravel(), yy.ravel()])
    Z = Z.reshape(xx.shape)
    plt.contourf(xx, yy, Z, alpha=0.3)
    plt.scatter(X[:, 0], X[:, 1], c=y, edgecolor='k')
    plt.xlabel("Sepal Length")
    plt.ylabel("Sepal Width")
    plt.title(title)
    plt.show()
```

```python
mlp_tanh = MLPClassifier(
    hidden_layer_sizes=(10,),
    activation='tanh',
    max_iter=1000,
    random_state=1
)
mlp_tanh.fit(X_train, y_train)
y_pred_tanh = mlp_tanh.predict(X_test)
print("Accuracy (Tanh):", accuracy_score(y_test, y_pred_tanh))
plot_decision_boundary(
    mlp_tanh, X_train, y_train,
    "Decision Boundary (Tanh Activation)"
)
plt.plot(mlp_tanh.loss_curve_)
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.title("Loss Curve (Tanh Activation)")
plt.grid(True)
plt.show()
```

```python
mlp_sigmoid = MLPClassifier(
    hidden_layer_sizes=(10,),
    activation='logistic',   
    max_iter=1000,
    random_state=1
)
mlp_sigmoid.fit(X_train, y_train)
y_pred_sigmoid = mlp_sigmoid.predict(X_test)
print("Accuracy (Sigmoid):", accuracy_score(y_test, y_pred_sigmoid))
plot_decision_boundary(
    mlp_sigmoid, X_train, y_train,
    "Decision Boundary (Sigmoid Activation)"
)
plt.plot(mlp_sigmoid.loss_curve_)
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.title("Loss Curve (Sigmoid Activation)")
plt.grid(True)
plt.show()
```

Exercise - 3: Design and implement an MLP to classify tumors as benign or malignant using the Breast Cancer Wisconsin dataset. Evaluate accuracy and plot the learning curve. Also plot the decision boundary.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.neural_network import MLPClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score
data = load_breast_cancer()
X = data.data[:, :2]
y = data.target
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=1
)
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
X_scaled = scaler.transform(X)
mlp = MLPClassifier(hidden_layer_sizes=(10, 5),
                    activation='relu',
                    solver='adam',
                    max_iter=2000,
                    random_state=1)
mlp.fit(X_train, y_train)
y_pred = mlp.predict(X_test)
print("Accuracy:", accuracy_score(y_test, y_pred))
plt.figure()
plt.plot(mlp.loss_curve_)
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.title("Learning Curve (Breast Cancer - MLP)")
plt.show()
x_min, x_max = X_scaled[:, 0].min() - 1, X_scaled[:, 0].max() + 1
y_min, y_max = X_scaled[:, 1].min() - 1, X_scaled[:, 1].max() + 1
xx, yy = np.meshgrid(np.linspace(x_min, x_max, 300), np.linspace(y_min, y_max, 300))
Z = mlp.predict(np.c_[xx.ravel(), yy.ravel()])
Z = Z.reshape(xx.shape)
plt.figure()
plt.contourf(xx, yy, Z, alpha=0.3)
plt.scatter(X_scaled[:, 0], X_scaled[:, 1], c=y)
plt.xlabel("Feature 1 (Scaled)")
plt.ylabel("Feature 2 (Scaled)")
plt.title("Decision Boundary (Breast Cancer - MLP)")
plt.show()
```

Exercise - 4: Implement an MLP to classify handwritten digits using the MNIST dataset and analyze 
the loss curve, decision boundary.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import fetch_openml
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import accuracy_score
mnist = fetch_openml('mnist_784', version=1)
X = mnist.data / 255.0
y = mnist.target.astype(int)
X = X[:20000]
y = y[:20000]
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=1
)
mlp = MLPClassifier(hidden_layer_sizes=(128,64),
                    activation='relu',
                    solver='adam',
                    max_iter=20,
                    random_state=1)
mlp.fit(X_train, y_train)
y_pred = mlp.predict(X_test)
print("Accuracy (Full Features):", accuracy_score(y_test, y_pred))
plt.figure()
plt.plot(mlp.loss_curve_)
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.title("Learning Curve (MNIST)")
plt.show()
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X)
X_train_p, X_test_p, y_train_p, y_test_p = train_test_split(
    X_pca, y, test_size=0.2, random_state=1
)
mlp_pca = MLPClassifier(hidden_layer_sizes=(50,),
                        activation='relu',
                        max_iter=50,
                        random_state=1)
mlp_pca.fit(X_train_p, y_train_p)
x_min, x_max = X_pca[:, 0].min()-1, X_pca[:, 0].max()+1
y_min, y_max = X_pca[:, 1].min()-1, X_pca[:, 1].max()+1
xx, yy = np.meshgrid(np.linspace(x_min, x_max, 300),
                     np.linspace(y_min, y_max, 300))
Z = mlp_pca.predict(np.c_[xx.ravel(), yy.ravel()])
Z = Z.reshape(xx.shape)
plt.figure()
plt.contourf(xx, yy, Z, alpha=0.3)
plt.scatter(X_pca[:, 0], X_pca[:, 1], c=y, s=5)
plt.title("Decision Boundary (MNIST - PCA Projection)")
plt.xlabel("PCA Component 1")
plt.ylabel("PCA Component 2")
plt.show()
```

Exercise - 5: Design an MLP to predict house prices using a given dataset. Plot training loss and comment on convergence.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neural_network import MLPRegressor
from sklearn.metrics import mean_squared_error, r2_score
data = fetch_california_housing()
X = data.data
y = data.target
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=1
)
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
mlp = MLPRegressor(hidden_layer_sizes=(100, 50),
                   activation='relu',
                   solver='adam',
                   learning_rate_init=0.001,
                   max_iter=200,
                   random_state=1)

mlp.fit(X_train, y_train)
y_pred = mlp.predict(X_test)
print("MSE:", mean_squared_error(y_test, y_pred))
print("R2 Score:", r2_score(y_test, y_pred))
plt.figure()
plt.plot(mlp.loss_curve_)
plt.xlabel("Epochs")
plt.ylabel("Training Loss (MSE)")
plt.title("Learning Curve (MLP - House Price Prediction)")
plt.show()
```

The training loss decreases gradually with epochs and stabilizes after several iterations, indicating proper convergence of the MLP model. The smooth decline suggests that the optimizer (Adam) successfully minimized the mean squared error without significant oscillations.
